```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, classification_report, confusion_matrix, roc_auc_score, roc_curve
```

```python
data_path_candidates = [
    Path.cwd() / 'Data' / 'Cleaned Dataset.csv',
    Path.cwd() / 'notebooks' / '..' / 'Data' / 'Cleaned Dataset.csv',
    Path.cwd() / '..' / 'Data' / 'Cleaned Dataset.csv',
    Path.cwd().parent / 'Data' / 'Cleaned Dataset.csv'
]
data_path = next((path for path in data_path_candidates if path.exists()), data_path_candidates[0])
data = pd.read_csv(data_path)

x = data.drop('cardio', axis = 1)
y = data['cardio']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

numeric_features = x_train.select_dtypes(include = "number").columns
categorical_features = x_train.select_dtypes(include = ["object", "category", "bool"]).columns

numeric_preprocessor = Pipeline(steps = [
    ('scaler', StandardScaler())
])

categorical_preprocessor = Pipeline(steps = [
    ('onehot', OneHotEncoder(handle_unknown = 'ignore'))
])

preprocessor = ColumnTransformer(transformers = [
    ('num', numeric_preprocessor, numeric_features),
    ('cat', categorical_preprocessor, categorical_features)
])
```

```python
model_1 = LogisticRegression(
    penalty=None,
    solver='lbfgs',
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

model_2 = LogisticRegression(
    penalty='l2',
    C=1.0,
    solver='lbfgs',
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

model_3 = LogisticRegression(
    penalty='l1',
    C=1.0,
    solver='liblinear',
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

model_4 = LogisticRegression(
    penalty='elasticnet',
    C=1.0,
    l1_ratio=0.5,
    solver='saga',
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

model_5 = LogisticRegression(
    penalty='elasticnet',
    C=1.0,
    l1_ratio=0.25,
    solver='saga',
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

model_6 = LogisticRegression(
    penalty='elasticnet',
    C=1.0,
    l1_ratio=0.75,
    solver='saga',
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)


logistic_models = [
    model_1, model_2, model_3,
    model_4, model_5, model_6
]

logistic_model_names = [
    'Logistic Regression (No Regularization)',
    'Logistic Regression (L2 Regularization)',
    'Logistic Regression (L1 Regularization)',
    'Logistic Regression (Elastic Net, l1_ratio=0.5)',
    'Logistic Regression (Elastic Net, l1_ratio=0.25)',
    'Logistic Regression (Elastic Net, l1_ratio=0.75)'
]

logistic_models_dict = dict(zip(
    logistic_model_names,
    logistic_models
))

logistic_pipelines = {
    name: Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    for name, model in logistic_models_dict.items()
}

model_7 = SGDClassifier(
    loss='log_loss',
    penalty=None,
    alpha=0.0001,
    n_iter_no_change=5,
    early_stopping=True,
    learning_rate='adaptive',
    tol=1e-3,
    eta0=0.01,
    max_iter=1000,
    class_weight='balanced',
    shuffle=True,
    random_state=42
)

model_8 = SGDClassifier(
    loss='log_loss',
    penalty='l2',
    alpha=0.0001,
    n_iter_no_change=5,
    early_stopping=True,
    learning_rate='adaptive',
    tol=1e-3,
    eta0=0.01,
    max_iter=1000,
    class_weight='balanced',
    shuffle=True,
    random_state=42
)

model_9 = SGDClassifier(
    loss='log_loss',
    penalty='l1',
    alpha=0.0001,
    n_iter_no_change=5,
    early_stopping=True,
    learning_rate='adaptive',
    tol=1e-3,
    eta0=0.01,
    max_iter=1000,
    class_weight='balanced',
    shuffle=True,
    random_state=42
)

model_10 = SGDClassifier(
    loss='log_loss',
    penalty='elasticnet',
    alpha=0.0001,
    l1_ratio=0.5,
    n_iter_no_change=5,
    early_stopping=True,
    learning_rate='adaptive',
    tol=1e-3,
    eta0=0.01,
    max_iter=1000,
    class_weight='balanced',
    shuffle=True,
    random_state=42
)

model_11 = SGDClassifier(
    loss='log_loss',
    penalty='elasticnet',
    alpha=0.0001,
    l1_ratio=0.25,
    n_iter_no_change=5,
    early_stopping=True,
    learning_rate='adaptive',
    tol=1e-3,
    eta0=0.01,
    max_iter=1000,
    class_weight='balanced',
    shuffle=True,
    random_state=42
)

model_12 = SGDClassifier(
    loss='log_loss',
    penalty='elasticnet',
    alpha=0.0001,
    l1_ratio=0.75,
    n_iter_no_change=5,
    early_stopping=True,
    learning_rate='adaptive',
    tol=1e-3,
    eta0=0.01,
    max_iter=1000,
    class_weight='balanced',
    shuffle=True,
    random_state=42
)


sgd_models = [
    model_7, model_8, model_9,
    model_10, model_11, model_12
]

sgd_model_names = [
    'SGD Classifier (No Regularization)',
    'SGD Classifier (L2 Regularization)',
    'SGD Classifier (L1 Regularization)',
    'SGD Classifier (Elastic Net, l1_ratio=0.5)',
    'SGD Classifier (Elastic Net, l1_ratio=0.25)',
    'SGD Classifier (Elastic Net, l1_ratio=0.75)'
]

sgd_models_dict = dict(zip(
    sgd_model_names,
    sgd_models
))

sgd_pipelines = {
    name: Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    for name, model in sgd_models_dict.items()
}
```

```python
stratified_kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = [
    'accuracy',
    'roc_auc',
    'f1',
    'precision',
    'recall'
]

logistic_results = []
sgd_results = []

for model_name, pipeline in logistic_pipelines.items():

    cross_val_results = cross_validate(
        pipeline,
        x_train,
        y_train,
        cv=stratified_kfold,
        scoring=scoring,
        return_train_score=True,
        error_score='raise'
    )

    logistic_results.append({
        'Model': model_name,
        'Train Accuracy': cross_val_results['train_accuracy'].mean(),
        'Mean Accuracy': cross_val_results['test_accuracy'].mean(),
        'Std Dev': cross_val_results['test_accuracy'].std(),
        'ROC AUC': cross_val_results['test_roc_auc'].mean(),
        'F1 Score': cross_val_results['test_f1'].mean(),
        'Precision': cross_val_results['test_precision'].mean(),
        'Recall': cross_val_results['test_recall'].mean()
    })

for model_name, pipeline in sgd_pipelines.items():

    cross_val_results = cross_validate(
        pipeline,
        x_train,
        y_train,
        cv=stratified_kfold,
        scoring=scoring,
        return_train_score=True,
        error_score='raise'
    )

    sgd_results.append({
        'Model': model_name,
        'Train Accuracy': cross_val_results['train_accuracy'].mean(),
        'Mean Accuracy': cross_val_results['test_accuracy'].mean(),
        'Std Dev': cross_val_results['test_accuracy'].std(),
        'ROC AUC': cross_val_results['test_roc_auc'].mean(),
        'F1 Score': cross_val_results['test_f1'].mean(),
        'Precision': cross_val_results['test_precision'].mean(),
        'Recall': cross_val_results['test_recall'].mean()
    })

lr_results_df = pd.DataFrame(logistic_results)
sgd_results_df = pd.DataFrame(sgd_results)

print("Logistic Regression Models Cross-Validation Results:")
print(lr_results_df.to_string(index=False))

print("\nSGD Classifier Models Cross-Validation Results:")
print(sgd_results_df.to_string(index=False))
```

```python
test_results = []

all_pipelines = {
    **logistic_pipelines,
    **sgd_pipelines
}

for model_name, pipeline in all_pipelines.items():

    pipeline.fit(x_train, y_train)

    y_pred = pipeline.predict(x_test)

    y_prob = pipeline.predict_proba(x_test)[:, 1]

    test_results.append({
        'Model': model_name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'ROC AUC': roc_auc_score(y_test, y_prob),
        'F1 Score': f1_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred)
    })

test_results_df = pd.DataFrame(test_results)

print("Test Set Evaluation Results:")
print(test_results_df.to_string(index=False))
```
